```python
from nltk.corpus import stopwords

def fraction_non_stopwords(text):
    """
    Return the fraction of alphabetic, lowercased tokens in `text`
    that are NOT English stopwords.
    """
    # Step 1: build the set of stopwords
    stop_words = set(stopwords.words('english'))

    # Step 2: filter `text` to keep only alphabetic tokens, lowercased
    words = [w.lower() for w in text if w.isalpha()]

    if len(words) == 0:
        return 0.0

    # Step 3: count how many of those are NOT in the stopword set
    non_stop = [w for w in words if w not in stop_words]

    return len(non_stop) / len(words)

# Step 4: return non-stop / total
# Test on a few NLTK book texts
from nltk.book import *

for name, txt in [('text1 (Moby Dick)', text1),
                  ('text2 (Sense & Sensibility)', text2),
                  ('text3 (Genesis)', text3),
                  ('text6 (Monty Python)', text6)]:
    frac = fraction_non_stopwords(txt)
    print(f"{name:<30} {frac:.3f}")
```
